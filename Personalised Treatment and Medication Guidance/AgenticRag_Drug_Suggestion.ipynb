```python
import pandas as pd

# Load CSV into DataFrame
df["rating"] = pd.to_numeric(df["rating"], errors="coerce")

# Sort by 'medical_condition' and 'rating' (descending)
df_sorted = df.sort_values(by=["medical_condition", "rating"], ascending=[True, False])

# Select top 7 drugs for each medical condition
top_drugs = (
    df_sorted.groupby("medical_condition")["drug_name"]
    .apply(lambda x: x.head(7).tolist())  # Get top 7 as a list
    .reset_index()
)

# Rename column to 'drugs'
top_drugs.rename(columns={"drug_name": "drugs"}, inplace=True)

# Display the new DataFrame
print(top_drugs)

# Save to a new CSV file
top_drugs.to_csv("top_7_drugs_per_condition.csv", index=False)
```

```python
import pandas as pd
from fpdf import FPDF

def convert_csv_to_pdf(csv_path, pdf_path):
    # Load CSV into Pandas DataFrame
    df = pd.read_csv(csv_path)

    # Ensure 'drugs' column is properly formatted
    df["drugs"] = df["drugs"].apply(lambda x: eval(x) if isinstance(x, str) else x)

    # Initialize PDF
    pdf = FPDF()
    pdf.set_auto_page_break(auto=True, margin=15)
    pdf.add_page()
    pdf.set_font("Arial", size=12)

    # Title
    pdf.set_font("Arial", style="B", size=16)
    pdf.cell(200, 10, "Medical Conditions and Their Treatments", ln=True, align="C")
    pdf.ln(10)  # Line break

    # Convert each row to a formatted sentence in the PDF
    pdf.set_font("Arial", size=12)
    for _, row in df.iterrows():
        condition = row["medical_condition"]
        drugs = ", ".join(row["drugs"])  # Convert list to comma-separated string
        text = f"{condition} can be cured by the following drugs: {drugs}."

        # Add text to PDF
        pdf.multi_cell(0, 10, text)
        pdf.ln(5)  # Extra space between records

    # Save PDF
    pdf.output(pdf_path)
    print(f"PDF saved successfully at: {pdf_path}")

# Example usage
csv_path = "/content/top_7_drugs_per_condition.csv"  # Replace with your CSV file path
pdf_path = "/content/medical_conditions.pdf"  # Output PDF file path
convert_csv_to_pdf(csv_path, pdf_path)
```

```python
import pandas as pd
from llama_index.core import Settings, KnowledgeGraphIndex, Document, StorageContext
from llama_index.embeddings.huggingface import HuggingFaceEmbedding
from llama_index.llms.groq import Groq

# Load CSV into Pandas DataFrame
csv_path = "/content/top_7_drugs_per_condition.csv"  # Replace with your actual file path
top_drugs = pd.read_csv(csv_path)

# Ensure 'drugs' column is properly formatted
top_drugs["drugs"] = top_drugs["drugs"].apply(lambda x: eval(x) if isinstance(x, str) else x)

# Convert DataFrame to formatted text
csv_text = "\n".join(
    f"{row['medical_condition']} can be cured by {', '.join(row['drugs'])}."
    for _, row in top_drugs.iterrows()
)

# Create a list of Document objects
documents = [Document(text=csv_text)]  # Wrap in a list

# Use HuggingFace's embedding model
embedding_model = HuggingFaceEmbedding(model_name="BAAI/bge-small-en-v1.5")

# Set embedding model
Settings.embed_model = embedding_model

# Initialize Groq LLM
llm = Groq(model="llama-3.3-70b-versatile")  # Replace with correct Groq model
Settings.llm = llm
Settings.chunk_size = 128

# Define storage context (use appropriate storage backend if needed)
storage_context = StorageContext.from_defaults()

# Build the Knowledge Graph Index
index = KnowledgeGraphIndex.from_documents(
    documents=documents,  # Ensure it's a list
    max_triplets_per_chunk=2,
    storage_context=storage_context,
)


```

```python
from llama_index.graph_stores.kuzu import KuzuGraphStore
import networkx as nx

# Assuming `index` is your KnowledgeGraphIndex instance
# Convert the knowledge graph to a NetworkX graph
nx_graph = index.get_networkx_graph()

# Iterate over nodes
print("Nodes in the Knowledge Graph:")
for node in nx_graph.nodes(data=True):
    print(node)

# Iterate over edges
print("\nEdges in the Knowledge Graph:")
cnt=0
for edge in nx_graph.edges(data=True):
  cnt = cnt+1
  print(edge)

print(cnt)
```
